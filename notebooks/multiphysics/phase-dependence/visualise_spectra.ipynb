# Visualise spectra: two QLT methods side by side
Overlays one spectrum (homodyne, heterodyne, position PSD or position CSD) computed by two quantum-linear-theory methods for the 2×2 system with unequal tweezer phases (`PHASE_DEPENDENT_xy_dict`, φ₁ = 0°, φ₂ = 30°, on the node). The first section shows one chosen pair; the second loops over every spectrum type and every numerical/analytic pair.

Runtime: about 1 min. The comparison plots are drawn by `levisim.multiphysics.visualise_two_spectrum_calculation_methods`.

# Visualise en masse

# Visualise spectra by evaluation order of the power series
Compares the homodyne, heterodyne, x₂ position PSD and x₂x₁ CSD from Power Series H, truncated at each evaluation order of the deranged determinant and adjugate, with LAM, for the FORTRAN pairs parameters. At full order the two agree; lower orders show what each order of hybridisation contributes.

Runtime: about 1-2 min for the default 2×2 system (`RUN_SYSTEMS` in the first code cell adds 5×1, 2×3, 7×1).

# START

# Visualise spectra at different evaluation orders
- Power Series H vs. LAM

```python
import numpy as np
from levisim import *
# not part of the shelf's public names, so import it explicitly
from levisim.multiphysics.utils.shelf import DEFAULT_Ω_model_equal_spacing_simple

# the systems to run: 2×2 takes about 1-2 min; add '5×1' (a few minutes), '2×3' (about 20 min)
# or '7×1' (about 3 h; the owner's timings, recorded in the cells below)
RUN_SYSTEMS = ['2×2']

def visualise_spectra_at_different_evaluation_orders(system):
    Nd = system.split('×')
    N, d = int(Nd[0]), int(Nd[1])

    # NOTE: default_toy_params is not used below (every system uses FORTRAN_toy_params, i.e. the pairs
    # model); for d == 1 it would also lack the couplings toy_gx, toy_gy
    default_toy_params = FORTRAN_toy_params
    if d == 1:
        default_toy_params = DEFAULT_Ω_model_equal_spacing_simple


    print('####################################################################################################')
    mtrap = MultiTrap(
        num_traps=N,
        Cartesian_axes_per_trap=d,
        ω_range=np.arange(0, 400, 0.5) * kHz,
        gas_kwargs=DEFAULT_gas_kwargs,
        cavity_kwargs=DEFAULT_cavity_kwargs,
        toy_params=FORTRAN_toy_params,
        generate_power_series=True,
    )
    evaluation_orders = list(range(N * d + 1))[::-1][:-1]
    for evaluation_order in evaluation_orders:
        print('----------------------------------------------------------------------------------------------------')
        mtrap.evaluation_order = evaluation_order
        print(f'{mtrap.evaluation_order=}')
        print(f'{len(mtrap.deranged_determinant_by_order[evaluation_order])=}')
        print(f'{len(mtrap.deranged_adjugate_by_order[evaluation_order]["x1x1"])=}')
        print(f'{len(mtrap.deranged_adjugate_by_order[evaluation_order]["x1x2"])=}')

        method_A = QuantumLinearTheoryMethod.LINEAR_AMPLIFIER_MODEL
        method_B = QuantumLinearTheoryMethod.POWER_SERIES_HYBRIDISATION

        visualise_two_spectrum_calculation_methods(
            mtrap, SpectrumType.HOMODYNE,
            method_A, method_B,
        )

        visualise_two_spectrum_calculation_methods(
            mtrap, SpectrumType.HETERODYNE,
            method_A, method_B,
        )

        visualise_two_spectrum_calculation_methods(
            mtrap, SpectrumType.POSITION_PSD,
            method_A, method_B,
            dim_label='x2',
        )

        visualise_two_spectrum_calculation_methods(
            mtrap, SpectrumType.POSITION_CSD,
            method_A, method_B,
            dim_label_A='x2',
            dim_label_B='x1',
        )
```

# 2×2 system

```python
# 39.7s
if '2×2' in RUN_SYSTEMS:
    visualise_spectra_at_different_evaluation_orders('2×2')
else:
    print("skipped: add '2×2' to RUN_SYSTEMS to run it")

# Power Series H spectral calculation durations (seconds):
# 24 (up to) 4th order determinant terms, 6 diagonal, 6 off-diagonal adjugate terms
#   Homodyne:       2.02
#   Heterodyne:     2.18
#   Position PSD:   0.95
#   Position CSD:   2.04
# 15 (up to) 4th order determinant terms, 6 diagonal, 6 off-diagonal adjugate terms
#   Homodyne:       2.13
#   Heterodyne:     1.72
#   Position PSD:   0.73
#   Position CSD:   1.93
# 7 (up to) 2nd order determinant terms, 4 diagonal, 3 off-diagonal adjugate terms
#   Homodyne:       0.77
#   Heterodyne:     0.78
#   Position PSD:   0.45
#   Position CSD:   0.85
# 1 (up to) 1st order determinant term, 1 diagonal, 1 off-diagonal adjugate terms
#   Homodyne:       0.63
#   Heterodyne:     0.34
#   Position PSD:   0.29
#   Position CSD:   0.27
```
